### Reducción de pdfs y annotations (ground truth)

```python
import os

# Folder paths (adjust if named differently)

pdf_folder = "../data/docile/pdfs"
annotations_folder = "../data/docile/annotations"

print("Synchronizing folders...")

# Collect base names of all existing PDFs
existing_pdfs = set()

for filename in os.listdir(pdf_folder):
    if filename.endswith(".pdf"):
        base_name = filename.replace(".pdf", "")
        existing_pdfs.add(base_name)

# Counters for final summary
jsons_checked = 0
jsons_deleted = 0

# Check each JSON file in the annotations folder
for filename in os.listdir(annotations_folder):
    if filename.endswith(".json"):
        jsons_checked += 1
        base_name = filename.replace(".json", "")
        # If the JSON name is not in the list of existing PDFs...
        if base_name not in existing_pdfs:
            orphan_json_path = os.path.join(annotations_folder, filename)
            # ...delete it
            try:
                os.remove(orphan_json_path)
                print(f"Deleted: {filename} (No matching PDF)")
                jsons_deleted += 1
            except Exception as e:
                print(f"Error deleting {filename}: {e}")

# Final summary
print("-" * 40)
print("Synchronization complete!")
print(f"PDFs found: {len(existing_pdfs)}")
print(f"JSONs checked: {jsons_checked}")
print(f"Orphan JSON files removed: {jsons_deleted}")
```

### Limpieza de annotations (ground truth)

```python
import json


def clean_annotations(json_path):

    with open(json_path, "r") as f:
        d = json.load(f)

    clean_d = {"line_item_extractions": d.get("line_item_extractions", [])}

    line_items = clean_d.get("line_item_extractions", [])
    clean_annotation = [
        {"fieldtype": item.get("fieldtype"), "text": item.get("text")}
        for item in line_items
    ]

    return {"extractions": clean_annotation}


clean = clean_annotations("../data/docile/annotations/fa29ed21c9044f66a5da45a1.json")
# print(clean)

for item in clean["extractions"]:
    print(item)
```
